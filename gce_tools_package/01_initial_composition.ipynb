# 1. Initial composition

Builds the starting guess for the solver from `run_toi561b.yaml`.

To change the planet or the initial guess, edit the YAML. To change the chemistry, edit (or copy) the network file in `networks/`. No code in this notebook needs changing.

```
from gce_tools import RunConfig, write_element_totals, write_initial

cfg = RunConfig.from_yaml("run_toi561b.yaml")
state = cfg.initial_state()
print(state.report(cfg.planet))
```

## Write solver inputs
`chem_input.dat` gets the element totals (`nSi`, `nMg`, ...). `initial.dat` gets the species fractions and phase moles.

```
WRITE = False   # set True to overwrite the files in cfg.solver.base_dir

if WRITE:
    write_element_totals(cfg.solver.path("chem_input"), state)
    write_initial(cfg.solver.path("initial"), state, update_only=True)
    print("Written to", cfg.solver.base_dir)
```
